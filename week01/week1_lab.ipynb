#### Fix 1 · Average grade
A student has three grades. The average should be 8.

````python
g1, g2, g3 = 7, 8, 9
average = g1 + g2 + g3 / 3
````

````python
assert average == 8, f"Expected 8, got {average}"
print("Fix 1 passed")
````

#### Fix 2 · Café receipt
Coffee 2.50 and croissant 1.80. Students get 10% off, then 24% VAT is added. The total should be 4.80.

````python
coffee = 2.50
croissant = 1.80
subtotal = coffee + croissant
discounted = subtotal * 0.10
total = round(discounted * 1.24, 2)
````

````python
assert total == 4.80, f"Expected 4.80, got {total}"
print("Fix 2 passed")
````

*What was wrong and why my fix works:*

#### Fix 3 · Age next year
The age arrives as text, the way input() would give it. Next year's age should be 26.

````python
age_text = "25"
next_year = age_text + 1
````

````python
assert next_year == 26, f"Expected 26, got {next_year}"
print("Fix 3 passed")
````

*What was wrong and why my fix works:*

#### Fix 4 · Minutes and seconds
135 seconds is 2 minutes and 15 seconds.

````python
total_seconds = 135
minutes = total_seconds / 60
seconds = total_seconds % 60
````

````python
assert minutes == 2, f"Expected 2 minutes, got {minutes}"
assert seconds == 15, f"Expected 15 seconds, got {seconds}"
print("Fix 4 passed")
````

*What was wrong and why my fix works:*
